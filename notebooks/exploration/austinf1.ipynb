```python
import fastf1
from fastf1 import plotting
import pandas as pd
import numpy as np
import os

import os, fastf1

CACHE_DIR = os.path.join(os.getcwd(), "cache_folder")  # absolute path
os.makedirs(CACHE_DIR, exist_ok=True)                  # create if missing
fastf1.Cache.enable_cache(CACHE_DIR)
print("Cache at:", CACHE_DIR)
```

```python
session = fastf1.get_session(2023, 'Austin', 'Qualifying')
session.load()
laps = session.laps

# Example: aggregate driver features
drivers = laps['Driver'].unique()
driver_stats = []
for d in drivers:
    drv_laps = laps.pick_driver(d)
    avg_lap = drv_laps['LapTime'].mean().total_seconds()
    best_lap = drv_laps['LapTime'].min().total_seconds()
    driver_stats.append({'Driver': d, 'AvgLap': avg_lap, 'BestLap': best_lap})
```

```python
import pandas as pd
df_drivers = pd.DataFrame(driver_stats)
df_drivers.sort_values('BestLap', inplace=True)
df_drivers.head()
```
